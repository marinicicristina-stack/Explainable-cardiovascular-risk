```python
import pandas as pd

# Column names of the Cleveland Heart Disease dataset
columns = [
    "age", "sex", "cp", "trestbps", "chol", "fbs",
    "restecg", "thalach", "exang", "oldpeak",
    "slope", "ca", "thal", "target"
]

# Load the dataset
df = pd.read_csv("heart_disease_cleveland.csv", names=columns)

# Display the first five rows
df.head()
```

```python
# Load the dataset # Charger le dataset
print("Number of patients:", df.shape[0])
print("Number of features:", df.shape[1])

# Display the first five rows # Afficher les 5 premières lignes
print("\nData types:")
print(df.dtypes)

# Check the distribution of the target variable # Vérifier les valeurs de la variable cible
print("\nTarget value distribution:")
print(df["target"].value_counts().sort_index())
```

```python
# Check for missing values represented by '?'# Rechercher les valeurs manquantes représentées par "?"
print("Number of '?' values per column:")
print((df == "?").sum())
```

```python
import numpy as np

# Replace '?' with NaN # Remplacer les "?" par NaN
df = df.replace("?", np.nan)

# Convert 'ca' and 'thal' columns to numeric values # Convertir ca et thal en valeurs numériques
df["ca"] = pd.to_numeric(df["ca"])
df["thal"] = pd.to_numeric(df["thal"])

# Check for missing values # Vérifier les valeurs manquantes
print("Missing values per column:")
print(df.isnull().sum())

print("\nTotal number of missing values:", df.isnull().sum().sum())
```

```python
# Remove rows containing missing values # Supprimer les lignes contenant des valeurs manquantes
df_clean = df.dropna().copy()

# Verify the cleaning results # Vérifier le résultat
print("Number of patients before cleaning:", len(df))
print("Number of patients after cleaning:", len(df_clean))
print("Remaining missing values:", df_clean.isnull().sum().sum())
```

```python
# Convert the target variable into binary classification # Transformer la cible en classification binaire
# 0 = No heart disease
# 1 = Heart disease

df_clean["target"] = (df_clean["target"] > 0).astype(int)

# Display the distribution of the binary target variable # Afficher la distribution de la nouvelle cible
print("Target distribution:")
print(df_clean["target"].value_counts())

print("\nTarget distribution (%):")
print(df_clean["target"].value_counts(normalize=True) * 100)
```
